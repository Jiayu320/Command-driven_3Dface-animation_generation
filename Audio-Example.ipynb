# Audio command driven 3D face animation generation

## Prerequisite

Need to convert origional auido file into wav format, you can use [this website](https://online-audio-converter.com/cn/) to do.

## Audio demo

### Audio command in English

```python
import torchaudio
from IPython import display
speech_signal_ori = 'audio/test_data/sad-shake-large-EN.wav'
audio_ori, rate_ori = torchaudio.load(str(speech_signal_ori))
display.Audio(audio_ori, rate=rate_ori)
```

### Audio command in Chinese

```python
speech_signal_ori = 'audio/test_data/happy-nod-large-CN.wav'
audio_ori, rate_ori = torchaudio.load(str(speech_signal_ori))
display.Audio(audio_ori, rate=rate_ori)
```

## Image demo

```python
display.Image(filename='TestSamples/real/me.jpg')
```

## Set running settings
### Audio command in English

**emotion_driven_function** ：Audio_driver <br>
**image_path**：TestSamples/real/me.jpg <br>
**audio_path**：'audio/test_data/sad-shake-large-EN.wav' <br>
**Language**：English<br>
**generate_frame**：40

```python
!python demos/demo_transfer.py --emotion_driven_function Audio_driver --image_path 'TestSamples/real/me.jpg' --audio_path 'audio/test_data/sad-shake-large-EN.wav' --language 'English' --generate_frame 40
```
