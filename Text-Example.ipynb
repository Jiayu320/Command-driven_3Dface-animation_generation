# Text command driven 3D face animation generation

## Image demo

```python
from IPython import display
display.Image(filename='TestSamples/real/me.jpg')
```

## Set running settings
### Text command in English

**emotion_driven_function** ：Text_driver<br>
**textInstrustion** ：Please make surprised movements and look to the left with large amplitude.<br>
**image_path**：TestSamples/real/me.jpg <br>
**Language**：English <br>
**generate_frame**：30

```python
!python demos/demo_transfer.py --emotion_driven_function Text_driver --textInstrustion 'Please make surprised movements and look to the left with large amplitude.' --image_path 'TestSamples/real/me.jpg' --language 'English' --generate_frame 30
```

#### Generation results

```python
display.Image(filename='./results/me_edit_surprised_left_large.gif')
```
